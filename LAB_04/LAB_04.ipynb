## 1. Local Speech-to-Text (Whisper)

```python
# Local STT Demo
stt = LocalWhisperSTT(model_name="base")

# Transcribe harvard.wav
harvard_path = settings.AUDIO_DIR / "harvard.wav"
harvard_path_resolved = harvard_path.resolve()

print(f"File exists: {harvard_path_resolved.exists()}")
print(f"File size: {harvard_path_resolved.stat().st_size / 1024 / 1024:.2f} MB")


result = stt.transcribe(str(harvard_path_resolved))
print("\nTranscription Result:")
print(f"  Text: {result['text']}")
print(f"  Language: {result['language']}")
```

## 2. Local Text-to-Speech (Piper TTS)

```python
# Local TTS Demo
try:
    tts = LocalPiperTTS(voice="en_US-lessac-medium", output_dir="output")
    output_file = tts.synthesize("Will Piper actually work? Idk this is a test.")
    print("Synthesis Result:")
    print(f"  Output file: {output_file}")
    print(f"  File size: {Path(output_file).stat().st_size / 1024:.1f} KB")
except Exception as e:
    print(f"Piper TTS unavailable: {e}")
    tts = None
```
