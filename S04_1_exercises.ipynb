# Bigrams

< s > I am Sam < /s >

< s > Sam I am < /s >

< s > I do not like green eggs and ham < /s >

### Calculate probabilities
P(I | < s >)

P(Sam | < s >)

P(am | I)

P(<\s> | Sam)

P(Sam | am)

P(do | I)

```python
import nltk
nltk.download('punkt')
```

```python
p1 = "I am Sam"
p2 = "Sam I am"
p3 = "I do not like green eggs and ham"

# Tokenize and add sentence markers
def tokenize(sent):
    return ["<s>"] + nltk.word_tokenize(sent) + ["</s>"]

tp1 = tokenize(p1)
tp2 = tokenize(p2)
tp3 = tokenize(p3)

# Create bigram corpus
p_all = tp1 + tp2 + tp3
all_bigrams = list(nltk.bigrams(p_all))
#print ("All Bigrams:", all_bigrams)

cfd = nltk.ConditionalFreqDist(all_bigrams)
cpd = nltk.ConditionalProbDist(cfd, nltk.MLEProbDist)

# Probabilities
print("P(I | <s>) = {:.2%}".format(cpd["<s>"].prob("I")))
print("P(Sam | <s>) = {:.2%}".format(cpd["<s>"].prob("Sam")))
print("P(am | I) = {:.2%}".format(cpd["I"].prob("am")))
print("P(</s> | Sam) = {:.2%}".format(cpd["Sam"].prob("</s>")))
print("P(Sam | am) = {:.2%}".format(cpd["am"].prob("Sam")))
print("P(do | I) = {:.2%}".format(cpd["I"].prob("do")))
```
